```python
# Convert Date column to datetime (if not already)
train['Date'] = pd.to_datetime(train['Date'])

# Add new date features
train['IsMonthStart'] = train['Date'].dt.is_month_start.astype(int)
train['IsMonthEnd'] = train['Date'].dt.is_month_end.astype(int)
train['IsQuarterStart'] = train['Date'].dt.is_quarter_start.astype(int)
train['IsQuarterEnd'] = train['Date'].dt.is_quarter_end.astype(int)

# Optional: Rolling average sales per store (if you have enough data)
train['Store_Avg_Sales'] = train.groupby('Store')['Sales'].transform('mean')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
import numpy as np
import xgboost as xgb

X_train, X_valid, y_train, y_valid = train_test_split(X, np.log1p(y), test_size=0.2, random_state=42)

xgb_model = xgb.XGBRegressor(
    objective='reg:squarederror',
    n_estimators=700,
    learning_rate=0.03,
    max_depth=10,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train, y_train)
y_pred_log = xgb_model.predict(X_valid)
y_pred = np.expm1(y_pred_log)

mae = mean_absolute_error(np.expm1(y_valid), y_pred)
print("🎯 New XGBoost Validation MAE:", mae)
```

```python
import streamlit as st
import pandas as pd
import numpy as np
import joblib
import os
import datetime

# Load latest saved model automatically
model_files = [f for f in os.listdir() if f.endswith(".pkl")]
latest_model = max(model_files, key=os.path.getctime)
model = joblib.load(latest_model)

st.title("💊 Rossmann Sales Forecasting Dashboard")

st.sidebar.header("Input Options")
option = st.sidebar.radio("Choose input type", ["Manual Input", "Upload CSV"])

# Manual input
if option == "Manual Input":
    store = st.number_input("Store ID", min_value=1, max_value=2000, step=1)
    promo = st.selectbox("Promo", [0, 1])
    school_holiday = st.selectbox("School Holiday", [0, 1])
    day = st.number_input("Day", 1, 31, 1)
    month = st.number_input("Month", 1, 12, 1)
    year = st.number_input("Year", 2013, 2025, 2025)
    day_of_week = st.number_input("Day of Week", 1, 7, 1)

    data = pd.DataFrame({
        'Store': [store],
        'Promo': [promo],
        'SchoolHoliday': [school_holiday],
        'Day': [day],
        'Month': [month],
        'Year': [year],
        'DayOfWeek': [day_of_week]
    })

    st.write("Input Data:", data)
    if st.button("Predict Sales"):
        pred_log = model.predict(data)
        pred = np.expm1(pred_log)
        st.success(f"💰 Predicted Sales: {pred[0]:.2f}")

# CSV upload
else:
    file = st.file_uploader("Upload CSV file", type=['csv'])
    if file is not None:
        df = pd.read_csv(file)
        st.write("Uploaded Data Preview:", df.head())

        preds = np.expm1(model.predict(df))
        df['Predicted_Sales'] = preds
        st.write("📈 Predictions:", df.head())

        csv = df.to_csv(index=False).encode('utf-8')
        st.download_button("Download Predictions", csv, "predicted_sales.csv", "text/csv")
```
